```python
import pandas as pd
import matplotlib.pyplot as plt

original = {
    "Class": [
        "crack",
        "damage",
        "erosion",
        "Metal",
        "mold",
        "0",
        "1",
        "2",
        "NoDamage",
        "Amber",
        "Green",
        "Red",
        "thick-dark",
        "thick-light",
        "thin-dark",
        "thin-light"
    ],
    "Count": [
        3717,
        1665,
        1665,
        67,
        938,
        47,
        47,
        46,
        752,
        752,
        752,
        752,
        233,
        233,
        233,
        233
    ]
}

df = pd.DataFrame(original)

df = df.sort_values("Count", ascending=True)

plt.figure(figsize=(9,6))

bars = plt.barh(
    df["Class"],
    df["Count"],
    color="#808080"
)

plt.xlabel("Number of Images")
plt.ylabel("Original Dataset Labels")
plt.title("Label Distribution Before Harmonization")

for bar in bars:
    plt.text(
        bar.get_width()+20,
        bar.get_y()+bar.get_height()/2,
        f"{int(bar.get_width())}",
        va="center"
    )

plt.tight_layout()
plt.show()
```

```python
harmonized = {
    "Category": [
        "Damage",
        "Wear",
        "Mold",
        "Asbestos",
        "No Damage"
    ],
    "Count": [
        8390,
        1732,
        1078,
        932,
        3008
    ]
}

df = pd.DataFrame(harmonized)

df = df.sort_values("Count", ascending=True)

plt.figure(figsize=(8,4))

bars = plt.barh(
    df["Category"],
    df["Count"],
    color="#ff0400"
)

plt.xlabel("Number of Images")
plt.ylabel("Harmonized Category")
plt.title("Label Distribution After Harmonization")

for bar in bars:
    plt.text(
        bar.get_width()+50,
        bar.get_y()+bar.get_height()/2,
        f"{int(bar.get_width())}",
        va="center"
    )

plt.tight_layout()
plt.show()
```
